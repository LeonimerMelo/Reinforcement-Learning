# Algoritmos Advantage Actor-Critic (A2C e A3C)

Os algoritmos Advantage Actor-Critic (A2C e A3C) são técnicas de aprendizado por reforço que combinam os métodos de política (actor) e valor (critic) de forma eficiente para treinar agentes em uma ampla variedade de tarefas. Eles são especialmente úteis em ambientes com alta dimensionalidade de estados e ações, e têm se mostrado eficazes em diversas aplicações, como jogos, robótica e controle de sistemas complexos.

**Conceitos Fundamentais:**

* **Actor:** Parte do algoritmo responsável por tomar as decisões, ou seja, escolher as ações a serem executadas em cada estado.
* **Critic:** Parte do algoritmo responsável por avaliar a qualidade das ações escolhidas pelo actor, estimando o valor de cada estado e a vantagem de cada ação.
* **Advantage:** A diferença entre o valor de uma ação e o valor médio de todas as ações possíveis em um determinado estado. Essa métrica ajuda a priorizar ações que levam a melhores resultados.

**Diferenças entre A2C e A3C:**

* **A2C (Advantage Actor-Critic):**
    * Utiliza um único processo para atualizar os parâmetros do actor e do critic.
    * É mais simples de implementar e geralmente apresenta um bom desempenho em muitos problemas.
    * É mais adequado para ambientes com um único agente.
* **A3C (Asynchronous Advantage Actor-Critic):**
    * Utiliza múltiplos processos (ou threads) para atualizar os parâmetros em paralelo.
    * Permite um treinamento mais rápido e estável, especialmente em ambientes complexos.
    * É mais adequado para ambientes com múltiplos agentes ou quando a simulação do ambiente é lenta.

**Funcionamento:**

1. **Inicialização:** Os parâmetros do actor e do critic são inicializados aleatoriamente.
2. **Interação com o ambiente:** O agente interage com o ambiente, coletando dados sobre os estados, ações e recompensas.
3. **Cálculo da vantagem:** A vantagem de cada ação é calculada com base na diferença entre o valor estimado do estado e o valor da ação.
4. **Atualização dos parâmetros:** Os parâmetros do actor e do critic são atualizados utilizando a técnica de gradiente descendente estocástico, com base na vantagem calculada.
5. **Repetição:** Os passos 2, 3 e 4 são repetidos por um número pré-definido de iterações ou até que um critério de parada seja atingido.

**Vantagens dos Algoritmos A2C e A3C:**

* **Convergência rápida:** Geralmente convergem para soluções de alta qualidade em um número menor de iterações.
* **Flexibilidade:** Podem ser aplicados a uma ampla variedade de problemas de aprendizado por reforço.
* **Estabilidade:** São relativamente estáveis e menos propensos a divergir durante o treinamento.
* **Paralelização:** A3C permite a utilização de múltiplos núcleos de processamento, acelerando significativamente o treinamento.

**Aplicações:**

* **Jogos:** Dominaram diversas plataformas, como o Atari e o OpenAI Gym.
* **Robótica:** Controle de robôs manipuladores, drones e veículos autônomos.
* **Sistemas de recomendação:** Personalização de conteúdo e produtos para usuários.
* **Finanças:** Previsão de séries temporais e otimização de portfólios.

**Considerações Finais:**

A escolha entre A2C e A3C depende da complexidade do problema, dos recursos computacionais disponíveis e das características do ambiente. Ambos os algoritmos são ferramentas poderosas para o desenvolvimento de agentes inteligentes e têm sido amplamente utilizados em diversas áreas da inteligência artificial.


Os algoritmos **Advantage Actor-Critic (A2C e A3C)** são melhorias sobre o algoritmo básico de aprendizado por reforço **REINFORCE**. Eles abordam algumas limitações do REINFORCE, como alta variância nos gradientes estimados e ineficiência na coleta de amostras. Abaixo explico como esses algoritmos funcionam e o que os diferencia:

---

## **REINFORCE: O Básico**
O REINFORCE utiliza **gradiente de política** para atualizar os parâmetros da política (rede neural). O gradiente é estimado como:

$$
\nabla_\theta J(\theta) = \mathbb{E}_{\tau} \left[ \sum_{t} \nabla_\theta \log \pi_\theta(a_t | s_t) G_t \right]
$$

- $ \pi_\theta(a_t | s_t) $: Probabilidade de tomar a ação $ a_t $ dado o estado $ s_t $.
- $ G_t $: Retorno acumulado futuro.
- Ele é **off-policy** e de alta variância, dificultando o treinamento eficiente.

---

## **Problemas do REINFORCE**
1. **Alta Variância**: A utilização do retorno $ G_t $ diretamente no gradiente pode levar a variações grandes entre as atualizações.
2. **Atualizações Ineficientes**: Ele depende de trajetórias completas para estimar o gradiente, o que torna a aprendizagem lenta.
3. **Uso Subótimo de Informação**: O REINFORCE não separa a qualidade das ações em relação a outras possíveis ações (ou seja, não avalia as ações relativas ao baseline).

---

## **Advantage Actor-Critic (A2C/A3C)**
Os algoritmos A2C e A3C introduzem várias melhorias para superar os problemas do REINFORCE:

### **1. Uso da Função Advantage**
Os algoritmos A2C/A3C utilizam a função **Advantage** $ A(s, a) $, que mede a diferença entre a recompensa esperada ao tomar uma ação específica e a recompensa média no estado:

$$
A(s, a) = Q(s, a) - V(s)
$$

- $ Q(s, a) $: Valor esperado ao tomar a ação $ a $ no estado $ s $.
- $ V(s) $: Valor esperado do estado $ s $ (baseline).

Substituir $ G_t $ pela vantagem $ A(s, a) $ reduz a variância porque leva em conta a qualidade relativa de uma ação.

O gradiente de política atualizado é:

$$
\nabla_\theta J(\theta) = \mathbb{E} \left[ \nabla_\theta \log \pi_\theta(a_t | s_t) A(s_t, a_t) \right]
$$

---

### **2. Atualizações Sincronizadas (A2C) ou Assíncronas (A3C)**
- **A2C (Advantage Actor-Critic):** Realiza atualizações de forma síncrona, coletando gradientes de múltiplos ambientes paralelos e agregando-os antes de atualizar a política.
- **A3C (Asynchronous Advantage Actor-Critic):** Usa múltiplos trabalhadores assíncronos, que exploram o ambiente de forma independente e atualizam a política centralizada de maneira paralela.

Esses métodos aumentam a eficiência computacional e melhoram a exploração, além de mitigar o problema de correlação entre as amostras.

---

### **3. Atualizações On-Policy**
O A2C/A3C ainda é um algoritmo **on-policy**, mas utiliza gradientes mais precisos ao reduzir a variância por meio da função Advantage e da exploração paralela.

---

### **4. Atualizações Menores e Mais Frequentes**
Ao contrário do REINFORCE, que usa trajetórias completas para calcular $ G_t $, o A2C/A3C usa **janelas temporais menores (n-steps)**, acumulando recompensas ao longo de $ n $ passos para estimar $ G_t $. Isso torna as atualizações mais rápidas e estáveis.

$$
G_t^{(n)} = \sum_{k=0}^{n-1} \gamma^k r_{t+k} + \gamma^n V(s_{t+n})
$$

---

### **Resumo das Melhores do A2C/A3C sobre REINFORCE**
1. **Redução da Variância**: Uso da função Advantage $ A(s, a) $.
2. **Exploração Mais Eficiente**: Ambientes paralelos ou assíncronos.
3. **Treinamento Mais Estável**: Atualizações mais frequentes e baseadas em múltiplas amostras.
4. **Maior Eficiência Computacional**: Utilização de múltiplos trabalhadores.

---



Aqui está um exemplo básico de implementação do algoritmo **Advantage Actor-Critic (A2C)** em **PyTorch**. Para simplificar, utilizamos um ambiente da biblioteca **Gym**, como `CartPole-v1`.

````python
# implementação do Advantage Actor-Critic (A2C) em PyTorch
import gymnasium as gym
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

# Configurações
ENV_NAME = "CartPole-v1"
GAMMA = 0.99
LR = 1e-3
N_STEPS = 5
MAX_EPISODES = 1000

# Rede Actor-Critic
class ActorCritic(nn.Module):
    def __init__(self, input_dim, action_dim):
        super(ActorCritic, self).__init__()
        self.shared_layers = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
        )
        # Cabeça da Política (Ator)
        self.actor = nn.Sequential(
            nn.Linear(128, action_dim),
            nn.Softmax(dim=-1)
        )
        # Cabeça do Crítico (Valor)
        self.critic = nn.Linear(128, 1)

    def forward(self, state):
        shared = self.shared_layers(state)
        policy = self.actor(shared)
        value = self.critic(shared)
        return policy, value

# Agente A2C
class A2CAgent:
    def __init__(self, env):
        self.env = env
        self.state_dim = env.observation_space.shape[0]
        self.action_dim = env.action_space.n
        self.model = ActorCritic(self.state_dim, self.action_dim)
        self.optimizer = optim.Adam(self.model.parameters(), lr=LR)

    def select_action(self, state):
        state = torch.FloatTensor(state).unsqueeze(0)
        policy, _ = self.model(state)
        action = torch.multinomial(policy, num_samples=1).item()
        return action

    def compute_advantage(self, rewards, values, next_value, dones):
        returns = []
        R = next_value
        for step in reversed(range(len(rewards))):
            R = rewards[step] + GAMMA * R * (1 - dones[step])
            returns.insert(0, R)
        returns = torch.tensor(returns)
        advantage = returns - values
        return returns, advantage

    def train(self, trajectory):
        states, actions, rewards, dones, next_state = trajectory
        states = torch.FloatTensor(states)
        actions = torch.LongTensor(actions)
        rewards = torch.FloatTensor(rewards)
        dones = torch.FloatTensor(dones)
        next_state = torch.FloatTensor(next_state).unsqueeze(0)

        # Predições do modelo
        _, next_value = self.model(next_state)
        policy, values = self.model(states)

        # Obtem a função Advantage
        values = values.squeeze()
        returns, advantage = self.compute_advantage(rewards, values, next_value.item(), dones)

        # Calcula perdas
        log_probs = torch.log(policy.gather(1, actions.unsqueeze(1)).squeeze())
        actor_loss = -(log_probs * advantage.detach()).mean()
        critic_loss = advantage.pow(2).mean()
        loss = actor_loss + critic_loss

        # Otimização
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

# Loop de Treinamento
def train_a2c():
    env = gym.make(ENV_NAME)
    agent = A2CAgent(env)

    for episode in range(MAX_EPISODES):
        state, _ = env.reset()  # Gymnasium retorna estado e info
        trajectory = {"states": [], "actions": [], "rewards": [], "dones": []}
        episode_reward = 0

        for t in range(1, 10000):
            action = agent.select_action(state)
            next_state, reward, done, truncated, _ = env.step(action)

            # Certifique-se de que o estado é uma lista ou ndarray
            state = np.array(state, dtype=np.float32).tolist()
            next_state = np.array(next_state, dtype=np.float32).tolist()

            # Armazena a experiência
            trajectory["states"].append(state)
            trajectory["actions"].append(action)
            trajectory["rewards"].append(reward)
            trajectory["dones"].append(done or truncated)
            state = next_state
            episode_reward += reward

            # Atualiza o agente a cada N passos
            if len(trajectory["rewards"]) == N_STEPS or done or truncated:
                trajectory["next_state"] = state
                agent.train((
                    trajectory["states"],
                    trajectory["actions"],
                    trajectory["rewards"],
                    trajectory["dones"],
                    trajectory["next_state"]
                ))
                trajectory = {"states": [], "actions": [], "rewards": [], "dones": []}

            if done or truncated:
                print(f"Episode {episode}, Reward: {episode_reward}")
                break

    env.close()

# Treinando o agente
if __name__ == "__main__":
    train_a2c()
````

### **Implementação do A2C com PyTorch**

Este exemplo segue os componentes principais do A2C:

1. **Ator-Crítico**: Rede neural que prediz a política ($ \pi(a|s) $) e o valor ($ V(s) $).
2. **Cálculo da Função Advantage**: Para reduzir a variância.
3. **Atualização dos Pesos**: Baseada no gradiente Advantage.

---

### **Como Funciona o Código**
1. **Rede Ator-Crítico**:
   - Camada compartilhada para aprender representações do estado.
   - Duas saídas:
     - A política (probabilidade de ações).
     - O valor esperado ($ V(s) $).

2. **Função Advantage**:
   - Calculada como $ A(s, a) = G_t - V(s) $, onde $ G_t $ é o retorno acumulado com desconto.

3. **Atualização dos Pesos**:
   - **Loss do Ator**: Maximiza o log da probabilidade ponderada pela vantagem.
   - **Loss do Crítico**: Minimiza a diferença quadrada entre o retorno e o valor estimado.

4. **Exploração Paralela** (não implementado neste exemplo):
   - A2C pode ser adaptado para rodar em múltiplos ambientes paralelos, mas este exemplo utiliza apenas um ambiente para simplificar.

O **A3C (Asynchronous Advantage Actor-Critic)** é uma extensão do A2C que permite treinar múltiplos agentes (ou "trabalhadores") de forma assíncrona em diferentes threads ou processos. Ele melhora a eficiência do treinamento e reduz a correlação entre as trajetórias coletadas pelos trabalhadores.

Abaixo, apresento uma implementação simplificada do A3C em Python usando o **PyTorch** e o **Gymnasium**, utilizando o módulo `multiprocessing` para lidar com os trabalhadores.

---

### **Código Completo do A3C**

````python
import gymnasium as gym
import torch
import torch.nn as nn
import torch.optim as optim
import torch.multiprocessing as mp
import numpy as np
from collections import deque

# Configurações globais
ENV_NAME = "CartPole-v1"
GAMMA = 0.99
LR = 1e-3
N_STEPS = 5
MAX_EPISODES = 1000
NUM_WORKERS = 4  # Número de threads/trabalhadores

# Rede Actor-Critic compartilhada entre os trabalhadores
class ActorCritic(nn.Module):
    def __init__(self, input_dim, action_dim):
        super(ActorCritic, self).__init__()
        self.shared_layers = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
        )
        self.actor = nn.Sequential(
            nn.Linear(128, action_dim),
            nn.Softmax(dim=-1)
        )
        self.critic = nn.Linear(128, 1)

    def forward(self, state):
        shared = self.shared_layers(state)
        policy = self.actor(shared)
        value = self.critic(shared)
        return policy, value

# Função auxiliar para calcular vantagem e retorno
def compute_advantage(rewards, values, next_value, dones):
    returns = []
    R = next_value
    for step in reversed(range(len(rewards))):
        R = rewards[step] + GAMMA * R * (1 - dones[step])
        returns.insert(0, R)
    returns = torch.tensor(returns)
    advantage = returns - values
    return returns, advantage

# Função principal de cada trabalhador
def worker(worker_id, global_model, optimizer, global_episode, results_queue):
    env = gym.make(ENV_NAME)
    local_model = ActorCritic(env.observation_space.shape[0], env.action_space.n)
    local_model.load_state_dict(global_model.state_dict())

    while global_episode.value < MAX_EPISODES:
        state, _ = env.reset()
        trajectory = {"states": [], "actions": [], "rewards": [], "dones": []}
        episode_reward = 0

        for t in range(1, 10000):
            state_tensor = torch.FloatTensor(state).unsqueeze(0)
            policy, value = local_model(state_tensor)
            action = torch.multinomial(policy, 1).item()

            next_state, reward, done, truncated, _ = env.step(action)
            episode_reward += reward

            # Armazenar trajetória
            # trajectory["states"].append(state)
            # trajectory["actions"].append(action)
            # trajectory["rewards"].append(reward)
            trajectory["states"].append(np.array(state, dtype=np.float32))
            trajectory["actions"].append(np.array(action, dtype=np.float32))
            trajectory["rewards"].append(np.array(reward, dtype=np.float32))
            trajectory["dones"].append(done or truncated)

            state = next_state
            if len(trajectory["rewards"]) == N_STEPS or done or truncated:
                next_state_tensor = torch.FloatTensor(next_state).unsqueeze(0)
                _, next_value = local_model(next_state_tensor)
                next_value = next_value.item() if not (done or truncated) else 0.0

                # Prepara tensores
                states = torch.FloatTensor(np.array(trajectory["states"]))
                actions = torch.LongTensor(np.array(trajectory["actions"]))
                rewards = torch.FloatTensor(np.array(trajectory["rewards"]))
                dones = torch.FloatTensor(trajectory["dones"])

                # Calcula vantagem e retornos
                policy, values = local_model(states)
                values = values.squeeze()
                returns, advantage = compute_advantage(rewards, values, next_value, dones)

                # Calcula perdas
                log_probs = torch.log(policy.gather(1, actions.unsqueeze(1)).squeeze())
                actor_loss = -(log_probs * advantage.detach()).mean()
                critic_loss = advantage.pow(2).mean()
                loss = actor_loss + critic_loss

                # Atualiza o modelo global
                optimizer.zero_grad()
                loss.backward()
                for global_param, local_param in zip(global_model.parameters(), local_model.parameters()):
                    global_param.grad = local_param.grad
                optimizer.step()

                # Atualiza o modelo local
                local_model.load_state_dict(global_model.state_dict())
                trajectory = {"states": [], "actions": [], "rewards": [], "dones": []}

            if done or truncated:
                with global_episode.get_lock():
                    global_episode.value += 1
                results_queue.put(episode_reward)
                print(f"Worker {worker_id}, Episode {global_episode.value}, Reward: {episode_reward}")
                break

    env.close()

# Função principal para treinar o A3C
def train_a3c():
    env = gym.make(ENV_NAME)
    global_model = ActorCritic(env.observation_space.shape[0], env.action_space.n)
    global_model.share_memory()  # Compartilhar modelo entre processos
    optimizer = optim.Adam(global_model.parameters(), lr=LR)
    global_episode = mp.Value("i", 0)
    results_queue = mp.Queue()

    workers = []
    for worker_id in range(NUM_WORKERS):
        worker_process = mp.Process(
            target=worker,
            args=(worker_id, global_model, optimizer, global_episode, results_queue)
        )
        workers.append(worker_process)

    for worker_process in workers:
        worker_process.start()

    for worker_process in workers:
        worker_process.join()

    # Coleta resultados
    results = []
    while not results_queue.empty():
        results.append(results_queue.get())

    print("Treinamento concluído!")
    print(f"Média de recompensa por episódio: {np.mean(results):.2f}")

if __name__ == "__main__":
    train_a3c()
````

---

### **Explicação do Código**
1. **Rede Actor-Critic Compartilhada**:
   - A classe `ActorCritic` é usada como a arquitetura para todos os trabalhadores. O modelo global é compartilhado entre os processos.
   
2. **Trabalhadores (`worker`)**:
   - Cada trabalhador executa o ambiente independentemente.
   - A cada `N_STEPS` ou quando o episódio termina, os trabalhadores:
     - Calculam vantagens e retornos.
     - Atualizam o modelo global (assincronicamente).
     - Sincronizam o modelo local com o modelo global.

3. **Processos Paralelos**:
   - A biblioteca `multiprocessing` é usada para executar vários trabalhadores em paralelo.
   - `global_model.share_memory()` permite que os processos compartilhem o mesmo modelo global.

4. **Sincronização**:
   - O gradiente é calculado localmente e aplicado ao modelo global.
   - Após cada atualização, os pesos do modelo global são sincronizados com o modelo local.

5. **Coleta de Recompensas**:
   - As recompensas de cada episódio são enviadas para uma fila (`results_queue`) para análise final.
