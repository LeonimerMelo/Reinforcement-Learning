#Introduction to Deep Q-Networks (DQN)

Deep Q-Networks são arquiteturas de aprendizado por reforço profundo que combinam o poder das redes neurais profundas com a eficiência do algoritmo Q-learning. Elas são amplamente utilizadas em problemas de tomada de decisão em ambientes sequenciais, como jogos e robótica.

Como funcionam?

 * **Rede Neural Profunda**: Uma rede neural profunda é treinada para aproximar a função Q, que estima a recompensa total futura a partir de um determinado estado e ação.
 * **Experiência Replay**: As interações do agente com o ambiente são armazenadas em um buffer de experiência. Isso permite que a rede seja treinada em uma variedade de estados e ações, tornando o aprendizado mais eficiente e estável.
 * **Aprendizado por Reforço**: A rede é treinada para minimizar a diferença entre as recompensas estimadas e as recompensas reais obtidas.

Exemplo Prático: Jogando Atari

Um dos exemplos mais famosos de aplicação de DQNs é em jogos Atari. Imagine um agente tentando jogar o jogo de Pong.
 * **Estados**: Cada frame do jogo representa um estado.
 * **Ações**: O agente pode mover a raquete para cima ou para baixo.
 * **Recompensas**: O agente recebe uma recompensa positiva por acertar a bola e uma recompensa negativa por perder o ponto.

A DQN é treinada para escolher a ação que maximiza a recompensa futura esperada a partir de cada estado. Com o tempo, a rede aprende a estratégia ótima para vencer o jogo.

**Aplicações**

 * Jogos: Jogar uma variedade de jogos, desde Atari até jogos mais complexos como StarCraft.
 * Robótica: Controlar robôs para realizar tarefas em ambientes complexos e dinâmicos.
 * Finanças: Tomar decisões de investimento.
 * Healthcare: Desenvolver tratamentos personalizados.

**Vantagens**

 * Flexibilidade: Podem ser aplicadas a uma ampla variedade de problemas.
 * Eficiência: Aprendem rapidamente e podem generalizar para novos estados.
 * Estabilidade: O mecanismo de experiência replay ajuda a estabilizar o treinamento.

**Desafios**

 * Tempo de treinamento: Podem levar muito tempo para treinar em ambientes complexos.
 * Hiperparâmetros: A escolha dos hiperparâmetros pode afetar significativamente o desempenho.
 * Exploração vs. Exploração: Encontrar o equilíbrio entre explorar novas ações e explorar as ações que já se mostraram eficazes.

**Em resumo**

DQNs são uma ferramenta poderosa para resolver problemas de tomada de decisão em ambientes complexos. Ao combinar o poder das redes neurais profundas com a eficiência do aprendizado por reforço, elas oferecem uma abordagem flexível e eficaz para uma ampla variedade de aplicações.

**Palavras-chave**: Deep Q-Networks, aprendizado por reforço, redes neurais profundas, inteligência artificial, jogos, robótica.

##Redes Neurais Profundas e Q-learning

O algoritmo de Q-learning, conforme já foi visto, possui muitas qualidades que permitem sua aplicação em diversos contextos do mundo real. Um elemento fundamental desse algoritmo é o uso da equação de Bellman para aprender a função Q. A equação de Bellman, conforme utilizada pelo algoritmo de Q-learning, permite a atualização dos valores Q a partir dos valores de estado-ação subsequentes. Isso possibilita que o algoritmo aprenda a cada passo, sem precisar esperar até que a trajetória seja concluída. Além disso, cada par estado-ação tem seus próprios valores armazenados em uma tabela de consulta, que salva e recupera os valores correspondentes.  

Sendo projetado dessa forma, o Q-learning converge para valores ótimos, desde que todos os pares estado-ação sejam amostrados repetidamente. Além disso, o método utiliza duas políticas: uma política de comportamento não gananciosa para coletar experiência do ambiente (por exemplo, uma política *ε-greedy*) e uma política-alvo gananciosa que segue o valor máximo de Q.  

Manter uma representação tabular dos valores pode ser contraindicado e, em alguns casos, prejudicial. Isso ocorre porque a maioria dos problemas apresenta um número muito alto de estados e ações. Por exemplo, imagens (mesmo as pequenas) possuem mais estados do que o número de átomos no universo. É fácil perceber que, nesse caso, o uso de tabelas se torna inviável. Além da memória infinita que seria necessária para armazenar tal tabela, apenas alguns estados seriam visitados mais de uma vez, tornando o aprendizado da função Q ou da função V extremamente difícil.  

Portanto, pode ser necessário generalizar entre estados. Nesse contexto, generalização significa que não estamos apenas interessados no valor exato de um estado, $V(s)$, mas também nos valores de estados semelhantes e próximos. Se um estado nunca foi visitado, podemos aproximá-lo pelo valor de um estado próximo a ele. De maneira geral, o conceito de generalização é extremamente importante em toda a área de aprendizado de máquina, incluindo aprendizado por reforço.

O conceito de generalização é fundamental em situações onde o agente não tem uma visão completa do ambiente. Nesse caso, o estado completo do ambiente estará oculto para o agente, que precisará tomar decisões com base apenas em uma representação restrita do ambiente. Isso é conhecido como **observação**.  

Por exemplo, imagine um agente humanoide que interage com o mundo real. Obviamente, ele não tem acesso ao estado completo do universo e de todos os átomos. Ele possui apenas uma visão limitada, ou seja, uma **observação**, que é percebida por seus sensores (como câmeras de vídeo). Por esse motivo, o agente humanoide precisa **generalizar** o que está acontecendo ao seu redor e agir de acordo com essa percepção.

### Aproximação de Função  

Agora que discutimos as principais limitações dos algoritmos tabulares e destacamos a necessidade de capacidades de generalização nos algoritmos de aprendizado por reforço (RL), precisamos lidar com as ferramentas que nos permitem superar essas restrições e abordar o problema da generalização.  

Podemos abandonar as tabelas e representar funções de valor por meio de um **aproximador de função**. A **aproximação de função** permite representar funções de valor em um domínio restrito usando apenas uma quantidade fixa de memória. A alocação de recursos depende apenas da função escolhida para aproximar o problema. A escolha do **aproximador de função** é sempre dependente da tarefa. Alguns exemplos incluem funções lineares, árvores de decisão, algoritmos de vizinhos mais próximos e redes neurais artificiais.  

Como era de se esperar, **redes neurais artificiais** são as mais utilizadas entre essas abordagens – não é por acaso que são amplamente adotadas em diversos algoritmos de RL. Em particular, redes neurais artificiais profundas, ou simplesmente ** deep neural networks (DNNs) - redes neurais profundas**, são empregadas devido à sua eficiência e capacidade de aprender representações automaticamente, criando uma estrutura hierárquica conforme aumentam as camadas ocultas da rede.  

Além disso, redes neurais profundas, especialmente as **convolutional neural networks (CNNs) - redes neurais convolucionais**, lidam excepcionalmente bem com imagens, conforme demonstrado por avanços recentes, principalmente em tarefas supervisionadas. No entanto, embora a maior parte dos estudos sobre redes neurais profundas tenha sido realizada no contexto de aprendizado supervisionado, sua integração em um framework de RL tem gerado resultados muito interessantes. Entretanto, como veremos a seguir, essa integração não é uma tarefa simples.

### Q-learning com Redes Neurais  

No Q-learning, uma rede neural profunda aprende um conjunto de pesos para aproximar a função de valor Q. Dessa forma, a função Q é parametrizada pelos pesos da rede e escrita da seguinte forma:

$$
Q(s, a; \theta),
$$
onde $\theta$ representa os pesos da rede neural.

Para adaptar o Q-learning com redes neurais profundas (essa combinação é chamada de **deep Q-learning**), é necessário criar uma função de perda (ou objetivo) a ser minimizada.

Como você deve se lembrar, a atualização do Q-learning tabular é dada por:

$$
Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]
$$

Aqui, $s'$ é o próximo estado e a atualização é feita **online**, em cada amostra coletada pela política de comportamento.

Ao integrar redes neurais no Q-learning, a fórmula para a atualização muda, pois a rede neural é usada para aproximar a função Q. Em vez de atualizar diretamente os valores em uma tabela, ajusta-se os pesos da rede neural para minimizar o erro entre a previsão da rede e o valor-alvo, com base na equação de Bellman.

Isso é feito calculando a diferença entre o valor estimado $Q(s, a; \theta)$ e o valor alvo $r + \gamma \max_{a'} Q(s', a'; \theta^-)$ e atualizando os pesos da rede de acordo com a função de perda. Essa função de perda pode ser a soma dos quadrados do erro (também conhecida como erro quadrático médio, MSE):

$$
L(\theta) = \left[ r + \gamma \max_{a'} Q(s', a'; \theta^-) - Q(s, a; \theta) \right]^2
$$

Dessa forma, a rede neural aprende a mapear estados e ações para valores Q apropriados, permitindo ao agente tomar decisões baseadas em uma representação contínua e mais compacta dos estados, em vez de depender de uma tabela enorme.

### Q-learning com Redes Neurais Profundas

Com a rede neural, nosso objetivo é otimizar os pesos $\theta$ para que a rede aprenda uma aproximação da função de valor Q ótima. No entanto, como não temos a função Q ótima, só podemos dar passos pequenos em direção a ela, minimizando o erro de Bellman de um único passo. Este passo é semelhante ao que fizemos no Q-learning tabular.

No entanto, no deep Q-learning, não atualizamos um único valor $Q(s, a)$. Em vez disso, tomamos o gradiente da função Q em relação aos parâmetros $\theta$:

$$
\nabla_\theta Q(s, a; \theta)
$$

Aqui, $\nabla_\theta Q(s, a; \theta)$ é a derivada parcial de $Q(s, a; \theta)$ em relação a $\theta$, e $\alpha$ é a taxa de aprendizado, que define o tamanho do passo a ser dado em direção ao gradiente.

Na prática, a transição suave do Q-learning tabular para o deep Q-learning não resulta em uma boa aproximação da função Q. A primeira melhoria envolve o uso do **Erro Quadrático Médio (MSE)** como uma função de perda, em vez do erro de Bellman. A segunda melhoria é a migração de uma iteração Q online para uma iteração Q em lotes (batch). Isso significa que os parâmetros da rede neural são atualizados usando múltiplas transições de uma vez (semelhante ao uso de um mini-lote de tamanho maior que 1 em configurações supervisionadas). Essas mudanças resultam na seguinte função de perda:

$$
L(\theta) = \mathbb{E}\left[\left( r + \gamma \max_{a'} Q(s', a'; \theta^-) - Q(s, a; \theta) \right)^2\right]
$$

Aqui, $Q(s, a; \theta)$ não é a verdadeira função de valor de ação, pois não a temos, mas sim o **valor alvo** $Q_{\text{target}}$. A função de perda é a média dos erros quadráticos, com base no valor $Q_{\text{target}}$.

Depois, os parâmetros da rede $\theta$ são atualizados usando o gradiente descendente aplicado à função de perda MSE:

$$
\theta \leftarrow \theta - \alpha \nabla_\theta L(\theta)
$$

Essas atualizações ajudam a rede neural a aproximar os valores $Q(s, a)$ ao longo do tempo, melhorando gradualmente a política do agente em um ambiente de aprendizado por reforço.

No Q-learning tabular, atualizamos diretamente o valor da função Q para um par estado-ação específico, $ Q(s, a) $, usando a equação de Bellman. Já no deep Q-learning, utilizamos uma rede neural para aproximar a função Q, cujos parâmetros são representados por $ \theta $. Em vez de atualizar um único valor, ajustamos $ \theta $ calculando o gradiente da função de perda.

A função de perda é geralmente definida como:

$$
L(\theta) = \mathbb{E} \left[ \left( y - Q(s, a; \theta) \right)^2 \right]
$$

onde $ y $ é o valor-alvo de Q:

$$
y = r + \gamma \max_{a'} Q(s', a'; \theta^-)
$$

Aqui:
- $ \theta $ representa os parâmetros atuais da rede Q.
- $ \theta^- $ representa os parâmetros da rede-alvo (atualizados periodicamente).
- $ r $ é a recompensa imediata.
- $ \gamma $ é o fator de desconto.
- $ s' $ é o próximo estado.

Ao minimizar essa função de perda por meio de descida do gradiente (usando otimizadores como Adam ou RMSprop), atualizamos $ \theta $ para melhorar a aproximação da função Q.

##Algoritmo Deep Q-Networks (DQN)
Aqui apresentamos o algoritmo Deep Q-Networks (DQN) proposto por Mnih et al. em 2013. Assim como o SARSA, o DQN é um algoritmo de diferença temporal (TD) baseado em valor que aproxima a função Q. A função Q aprendida é então usada por um agente para selecionar ações. O DQN é aplicável apenas a ambientes com espaços de ação discretos. No entanto, o DQN aprende uma função Q diferente em comparação ao SARSA — a função Q ótima em vez da função Q para a política atual. Essa pequena, mas crucial mudança melhora a estabilidade e a velocidade do aprendizado.

O DQN aprende a função Q ótima ao analisar a equação de Bellman para o DQN. Uma implicação importante é que isso torna o DQN um algoritmo off-policy. Esta função Q ótima não depende da política de coleta de dados, o que significa que o DQN pode aprender a partir de experiências coletadas por qualquer agente.

Em teoria, qualquer política pode ser usada para gerar dados de treinamento para o DQN. Na prática, no entanto, algumas políticas são mais apropriadas do que outras.

Basta uma pequena mudança na atualização da função Q para transformar o SARSA em DQN, mas o resultado é um algoritmo que aborda muitas das limitações do SARSA. O DQN possibilita decorrelacionar e reutilizar experiências ao amostrar lotes aleatórios de uma grande memória de replay de experiência, e permite múltiplas atualizações de parâmetros usando o mesmo lote. Essas ideias melhoram significativamente a eficiência do DQN em relação ao SARSA.




### **Diferença entre DQN e SARSA**
Tanto o **Deep Q-Network (DQN)** quanto o **SARSA** aprendem a função $ Q $ utilizando aprendizado por diferença temporal (TD learning). No entanto, os dois algoritmos diferem na forma como constroem o valor-alvo $ Q^\pi_{\text{tar}}(s, a) $, que é usado para atualizar a estimativa da função $ Q $.

---

#### **Equação de Bellman para DQN e SARSA**
A seguir, estão as equações de Bellman para DQN e SARSA:

1. **DQN:**
   $$
   Q^\pi_{\text{DQN}}(s, a) \approx r + \gamma \max_{a'} Q^\pi(s', a')
   $$
   - O DQN utiliza o valor máximo $ \max_{a'} Q^\pi(s', a') $ entre todas as ações possíveis no próximo estado $ s' $ para construir o valor-alvo.
   - Isso significa que o DQN escolhe sempre a melhor ação estimada, independentemente da política atual (*off-policy*).

2. **SARSA:**
   $$
   Q^\pi_{\text{SARSA}}(s, a) \approx r + \gamma Q^\pi(s', a')
   $$
   - O SARSA utiliza o valor da função $ Q $ para a ação específica $ a' $ selecionada pela política atual no próximo estado $ s' $.
   - Aqui, o valor-alvo depende diretamente da política sendo seguida (*on-policy*).

---

#### **Construção do Valor-Alvo $ Q^\pi_{\text{tar}}(s, a) $**
As fórmulas para a construção do valor-alvo em cada algoritmo são as seguintes:

1. **DQN:**
   $$
   Q^\pi_{\text{tar:DQN}}(s, a) = r + \gamma \max_{a'} Q^\pi(s', a')
   $$
   - O DQN escolhe a ação que maximiza a recompensa esperada no próximo estado, independentemente da política usada para explorar.

2. **SARSA:**
   $$
   Q^\pi_{\text{tar:SARSA}}(s, a) = r + \gamma Q^\pi(s', a')
   $$
   - O SARSA atualiza o valor $ Q^\pi(s, a) $ com base na ação específica $ a' $ que foi efetivamente escolhida pela política.

---

### **DQN é Off-Policy e SARSA é On-Policy**

#### **DQN (Deep Q-Network):**
No **DQN**, em vez de usar a ação $ a' $ realmente tomada no próximo estado $ s' $ para estimar $ Q^\pi_{\text{tar}}(s, a) $, ele utiliza o **maior valor $ Q $** entre todas as ações possíveis no estado $ s' $:
$$
Q^\pi_{\text{tar:DQN}}(s, a) = r + \gamma \max_{a'} Q^\pi(s', a')
$$

- O valor $ Q $ usado no estado $ s' $ não depende da política utilizada para coletar as experiências.
- Como a recompensa $ r $ e o próximo estado $ s' $ são gerados pelo ambiente com base no estado atual $ s $ e na ação $ a $, **nenhuma parte da estimativa $ Q^\pi_{\text{tar:DQN}}(s, a) $** depende da política de coleta de dados.
- Isso torna o DQN um **algoritmo off-policy**, pois a função $ Q $ sendo aprendida é independente da política seguida para agir no ambiente e coletar experiências.

---

#### **SARSA:**
Por outro lado, no **SARSA**, a estimativa $ Q^\pi_{\text{tar:SARSA}}(s, a) $ utiliza a **ação $ a' $** realmente tomada pela política atual no estado $ s' $:
$$
Q^\pi_{\text{tar:SARSA}}(s, a) = r + \gamma Q^\pi(s', a')
$$

- Nesse caso, a estimativa do valor $ Q $ depende diretamente da política usada para explorar o ambiente e escolher a ação $ a' $ no próximo estado.
- Isso torna o SARSA um **algoritmo on-policy**, pois ele considera explicitamente a política seguida para coletar experiências durante o aprendizado.

---

### **Resumo da Diferença:**
- **DQN:** Off-policy porque aprende $ Q $ considerando a melhor ação estimada ($ \max_{a'} Q^\pi(s', a') $), independentemente das ações efetivamente tomadas durante a exploração.
- **SARSA:** On-policy porque aprende $ Q $ com base na ação $ a' $ realmente selecionada pela política atual, tornando-o dependente das experiências coletadas.

Essa diferença entre DQN e SARSA é fundamental, pois afeta como os dois algoritmos lidam com o processo de aprendizado em ambientes dinâmicos. Enquanto o DQN tende a ser mais eficiente para aprender a política ótima, o SARSA é mais conservador, sendo útil em cenários onde a política exploratória precisa ser segura ou consistente com o comportamento observado.

### **Seleção de Ações no DQN**
Embora o DQN seja um algoritmo **off-policy**, a forma como o agente DQN coleta experiências ainda é relevante. Existem dois fatores importantes a serem considerados:

#### **1. Tradeoff exploration-exploitation**
- Um agente enfrenta o dilema entre **exploration** e **exploitation** (exploration-exploitation tradeoff).  
- No início do treinamento, o agente deve explorar rapidamente o espaço de estados e ações para aumentar as chances de descobrir boas formas de agir no ambiente.  
- À medida que o treinamento avança e o agente aprende mais, ele deve reduzir gradualmente a taxa de exploração e passar mais tempo **exploitation** o que já aprendeu.  
- Isso melhora a eficiência do treinamento, pois o agente se concentra em ações melhores.

Esse é um problema clássico em aprendizado por reforço, onde o agente precisa equilibrar dois objetivos conflitantes:

1. **Exploration**: Testar novas ações ou estratégias para descobrir informações sobre o ambiente.  
   - Por exemplo, um agente pode tentar ações que nunca realizou antes para entender se elas podem levar a boas recompensas.  
   - Isso é especialmente importante no início do treinamento, quando o agente não tem muito conhecimento sobre o ambiente.

2. **Exploitation**: Usar o conhecimento atual para escolher as ações que provavelmente maximizam as recompensas.  
   - Por exemplo, com base nos valores $ Q(s, a) $ aprendidos, o agente pode escolher a melhor ação em cada estado conhecido.

#### Como o DQN resolve esse tradeoff?
O DQN geralmente usa a estratégia **ε-greedy** para equilibrar exploration e exploitation:
- O agente escolhe uma **ação aleatória** com probabilidade $ \epsilon $ (exploration).  
- Com probabilidade $ 1 - \epsilon $, escolhe a **melhor ação conhecida** com base no valor $ Q(s, a) $ (exploitation).

#### **2. Espaço de Estados e Ações Muito Grande**
- Quando o espaço de estados e ações é muito grande, seja porque consiste em valores contínuos ou em valores discretos com alta dimensionalidade, será **impraticável** experimentar todas as combinações $ (s, a) $, mesmo uma única vez.  
- Nesse caso, os valores $ Q $ associados a pares $ (s, a) $ não visitados podem ser **tão imprecisos quanto uma suposição aleatória**.  



##Deep Q-Learning
O **Deep Q-Network (DQN)** é uma abordagem revolucionária que combina Q-learning, um método clássico de aprendizado por reforço, com redes neurais profundas. Essa combinação permite que agentes aprendam a tomar decisões em ambientes complexos diretamente a partir de entradas brutas, como imagens de pixels.  

Antes do DQN, a aplicação de Q-learning em problemas com grandes espaços de estado enfrentava grandes limitações devido ao uso de tabelas de valores Q, que não eram viáveis para ambientes contínuos ou de alta dimensionalidade. O DQN resolve esse problema ao aproximar a função de valor-ação $Q(s, a)$ com redes neurais profundas, viabilizando o aprendizado em problemas complexos, como jogos de Atari, onde o estado é representado por pixels da tela.  

No entanto, essa integração apresentou desafios de estabilidade e convergência. Para resolver essas questões, o DQN introduziu duas inovações principais:
1. **Replay Buffer**: Um mecanismo de armazenamento de experiências que desacopla a correlação temporal entre amostras e melhora a eficiência do aprendizado.  
2. **Rede Alvo (Target Network)**: Uma cópia estática da rede Q, que é atualizada periodicamente para garantir estabilidade durante o treinamento.

Com essas inovações, o DQN alcançou avanços significativos, permitindo que agentes superassem desempenhos humanos em diversos jogos de Atari. Essa abordagem foi um marco no aprendizado por reforço profundo e inspirou o desenvolvimento de algoritmos ainda mais avançados.

Para alcançar a tomada de decisão ponta a ponta em problemas complexos com entrada de pixels brutos, o DQN combina Q-learning com aprendizado profundo, utilizando duas ideias-chave para resolver o problema de instabilidade e alcança progressos significativos em jogos do Atari.  

###Replay Buffer
A primeira ideia é conhecida como **replay buffer**, um mecanismo inspirado biologicamente chamado de "replay de experiências" (McClelland et al. 1995; O’Neill et al. 2010; Lin 1993). A cada passo de tempo $t$, o DQN armazena a experiência do agente $(S_t, A_t, R_t, S_{t+1})$ no replay buffer e, em seguida, extrai uma mini-batch de amostras desse buffer uniformemente para aplicar a atualização do Q-learning.  

O replay buffer possui várias vantagens em relação à iteração Q ajustada:  
1. **Eficiência de Dados**: Cada experiência pode ser reutilizada para aprender a função Q, o que aumenta a eficiência dos dados.  
2. **Redução de Variância**: Sem o replay buffer, como na iteração Q ajustada, as amostras de mini-batches são coletadas consecutivamente, ou seja, são altamente correlacionadas, o que aumenta a variância das atualizações.  
3. **Aprendizado Suavizado**: O replay buffer evita que as amostras usadas para o treinamento sejam determinadas pelos parâmetros anteriores, suavizando o aprendizado e reduzindo oscilações ou divergências nos parâmetros.  

Na prática, apenas os últimos $N$ pares de experiência são armazenados no replay buffer para economizar memória.

### Replay Buffer: Replay de Experiência no DQN

O Replay de Experiência aborda duas ineficiências importantes encontradas em algoritmos de aprendizado por reforço *on-policy*, como o SARSA. Vamos detalhar:

1. **Uso Limitado dos Dados**:  
   - Algoritmos *on-policy* utilizam apenas os dados coletados pela política atual para aprender. Isso significa que cada experiência (estado, ação, recompensa, próximo estado) é usada apenas uma vez para atualizar o modelo.  
   - Isso é especialmente problemático ao usar métodos como redes neurais, que aprendem por meio de descida de gradiente. A descida de gradiente funciona melhor com atualizações pequenas e incrementais nos parâmetros. No entanto, em algumas situações, as previsões da rede podem precisar de ajustes significativos – por exemplo, quando o valor Q previsto pela rede $ \hat{Q}^{\pi}(s, a_1) $ é muito diferente do valor Q real $ Q^{\pi}(s, a_1) $. Métodos *on-policy* não têm mecanismo para revisitar e aprender repetidamente com essas experiências.

2. **Atualizações Subótimas dos Parâmetros**:  
   - Quando há uma grande diferença entre os valores Q previstos e reais, são necessárias várias atualizações para extrair todas as informações úteis da experiência. Infelizmente, métodos *on-policy* descartam os dados após um único uso, tornando impossível refinar os parâmetros da rede com base nessa experiência.

Armazenando experiências em um buffer de replay, o DQN permite que o modelo aprenda com experiências passadas várias vezes. Isso garante um uso mais eficiente dos dados e possibilita atualizações de parâmetros mais eficazes, melhorando significativamente a eficiência de amostragem em comparação com os métodos tradicionais *on-policy*.

### Replay de Experiência: Detalhes e Benefícios

Além do problema de uso limitado dos dados, os algoritmos *on-policy* enfrentam outro desafio:

 **Altamente Correlacionados**:  
   - As experiências usadas para treinar algoritmos *on-policy* tendem a ser altamente correlacionadas. Isso ocorre porque os dados utilizados para calcular uma única atualização de parâmetros geralmente vêm de um único episódio, onde os estados futuros e recompensas dependem de estados e ações anteriores. Essa correlação pode gerar uma alta variância nas atualizações dos parâmetros.

---

### O Papel do *Replay* de Experiência

Algoritmos *off-policy*, como o DQN, superam essas limitações ao não descartar experiências após o uso. Long-Ji Lin, em 1992, propôs uma extensão ao *Q-learning* chamada *Replay* de Experiência, para abordar essa questão. Ele observou que o aprendizado TD (*Temporal Difference*) pode ser lento devido ao processo de tentativa e erro necessário para coletar dados e à propagação da informação ao longo do tempo. Para acelerar o aprendizado TD, é necessário:  
   - Aumentar a eficiência no processo de atribuição de crédito.  
   - Ou encurtar o processo de tentativa e erro.

O *Replay* de Experiência foca no segundo ponto, permitindo o reaproveitamento de experiências.

---

### Como Funciona o *Replay* de Experiência

- **Memória de Replay**:  
  Armazena as $ k $ experiências mais recentes coletadas pelo agente. Quando a memória está cheia, as experiências mais antigas são descartadas para abrir espaço para as novas.  

- **Treinamento com Lotes Aleatórios**:  
  Toda vez que o agente treina, um ou mais lotes de dados são amostrados aleatoriamente (de forma uniforme) da memória de replay. Esses lotes são usados para atualizar os parâmetros da rede de função Q.  

- **Tamanho da Memória e dos Lotes**:  
  - O tamanho da memória $ k $ é geralmente grande, variando de 10.000 a 1.000.000 experiências.  
  - O tamanho dos lotes amostrados é muito menor, variando de 32 a 2.048 experiências por lote.  

---

### Benefícios do *Replay* de Experiência

1. **Decorrelação das Experiências**:  
   - Como cada lote contém experiências de diferentes episódios e políticas, as experiências usadas no treinamento são menos correlacionadas. Isso reduz a variância das atualizações dos parâmetros e estabiliza o processo de treinamento.

2. **Eficiência no Uso de Dados**:  
   - A memória deve ser grande o suficiente para conter várias experiências de diferentes episódios, mas pequena o suficiente para garantir que cada experiência seja reutilizada várias vezes antes de ser descartada. Isso torna o aprendizado mais eficiente.

3. **Foco em Experiências Recentes**:  
   - Descartar as experiências mais antigas é crucial. À medida que o agente aprende, a distribuição de pares estado-ação $ (s, a) $ muda. Experiências antigas tornam-se menos úteis, já que o agente tende a não visitar mais esses estados.  
   - Ao armazenar apenas as $ k $ experiências mais recentes, o agente se concentra em experiências mais relevantes, tornando o aprendizado mais eficaz dentro dos limites de tempo e recursos computacionais disponíveis.  

---

Essa abordagem é um dos principais motivos pelos quais o DQN é mais eficiente e estável no treinamento, especialmente em ambientes complexos.

### Rede Alvo (Target Network)
Para melhorar a estabilidade do Q-learning com redes neurais, o DQN introduz a **rede alvo (target network)**, que resolve problemas de instabilidade e oscilações.  
- Em vez de usar a Q-network atual para calcular os alvos do Q-learning (o que pode causar loops de feedback e divergência), a **rede alvo** é usada para gerar esses alvos.  
- A **rede alvo** é uma cópia da Q-network, mas é atualizada com menos frequência:
  - **Atualização rígida (hard update)**: Copia os pesos da Q-network para a rede alvo a cada $C$ passos.
  - **Atualização suave (soft update)**: Atualiza gradualmente a rede alvo usando uma média exponencialmente decrescente dos pesos da Q-network.

Essa atualização atrasada ajuda a estabilizar o treinamento, suavizando as mudanças no alvo do Q-learning e reduzindo superestimações e oscilações. Por exemplo:
- Se atualizar $Q(s_t, a_t)$ causa um aumento em $Q(s_{t+1}, a)$ para todas as ações $a$ devido a semelhanças entre os estados $s_t$ e $s_{t+1}$, a rede alvo garante que essas atualizações não sejam excessivamente agressivas, evitando superestimações.

---

### Usando redes-alvo  

Uma maneira muito direta de tornar os valores-alvo mais estacionários é ter uma rede separada que possamos manter fixa por múltiplos passos e reservá-la para calcular alvos mais estáveis. A rede com essa finalidade no DQN é chamada de **rede-alvo** (**target network**).  

### Como funciona:  
1. **Duas redes**:  
   - **Rede online** ($\theta$): É a rede principal que está sendo treinada.  
   - **Rede-alvo** ($\theta^-$): É uma cópia da rede online, mas atualizada com menos frequência.  

2. **Atualização dos alvos**:  
   - Os valores-alvo de Q são calculados usando a rede-alvo:  
     $$
     y = r + \gamma \max_{a'} Q(s', a'; \theta^-)
     $$
   - Em vez de usar a rede online (que muda constantemente) para calcular $ Q(s', a') $, a rede-alvo fornece uma referência mais estável.  

3. **Atualizações periódicas**:  
   - Os parâmetros da rede-alvo ($\theta^-$) são **atualizados periodicamente**, copiando os pesos da rede online ($\theta$), ou seja:  
     $$
     \theta^- \leftarrow \theta
     $$
   - Isso evita um ciclo de feedback onde uma função Q que muda rapidamente poderia desestabilizar o aprendizado.  

<img src='https://drive.google.com/uc?id=19ifhGBaz-0neI9l6VVcSZ_dTgOeRDnlS' width=700>

Isso explica a importância de usar uma *target network* em algoritmos de aprendizado por reforço profundo baseados em valor. O principal benefício é mitigar a instabilidade causada pela atualização simultânea das previsões e dos alvos, um problema conhecido como *chasing your own tail*.  

A ideia é transformar o aprendizado em uma sequência de pequenos problemas supervisionados, onde os alvos são mantidos constantes por um número fixo de passos. Isso não garante convergência para os valores ótimos, especialmente devido à aproximação por funções não lineares, mas melhora a estabilidade do aprendizado e reduz significativamente o risco de divergência—um problema frequente nesses métodos.  

<img src='https://drive.google.com/uc?id=1meoeO9MYGY-oKg-ysPVYh4PF-LkdQtuK' width=600>

### Benefícios:  
✅ **Reduz a variância**: Ao manter a rede-alvo fixa por um tempo, os valores-alvo se tornam mais estáveis.  
✅ **Melhora a convergência**: O treinamento tem menos chances de divergir devido a oscilações nos valores de Q.  
✅ **Funciona bem com replay de experiência**: Como experiências passadas são reutilizadas, um alvo estável evita que a rede aprenda valores inconsistentes.

---

### **Melhorias no Treinamento do DQN**  
O DQN introduz componentes e técnicas adicionais para refinar o processo de treinamento:  
1. **Representação de Histórico com Comprimento Fixo**:  
   - Alimentar um histórico de comprimento arbitrário na rede neural é computacionalmente inviável.  
   - Em vez disso, uma função $\phi$ comprime o histórico em uma representação de comprimento fixo empilhando os 4 quadros mais recentes (quadro atual + 3 anteriores). Essa abordagem ajuda o agente a capturar informações temporais, como o movimento de objetos.

2. **Pré-processamento**:  
   - Os quadros brutos são redimensionados para $84 \times 84$ pixels e convertidos para escala de cinza para reduzir a complexidade computacional.  
   - Os 4 quadros mais recentes são empilhados para formar a entrada da rede neural.

3. **Arquitetura da Rede Neural**:  
   - A rede consiste em:  
     - **Três camadas convolucionais**: Extraem representações espaciais e de características dos quadros de entrada.  
     - **Duas camadas totalmente conectadas**: Mapeiam as características extraídas para uma representação adequada para tomada de decisão.  
     - **Camada de saída**: Produz um único valor $Q$ para cada ação válida.

##Outro exemplo de algoritmo DQN
#### **1. Hiperparâmetros**:
- **Capacidade do Replay Buffer ($N$)**: Número máximo de transições armazenadas no buffer de repetição.  
- **Fator de Desconto ($\gamma$)**: Fator que desconta recompensas futuras.  
- **Passos de Atualização ($C$)**: Intervalo para atualizar a rede alvo ($Q$).  
- **Fator $\epsilon$-Greedy ($\epsilon$)**: Parâmetro que controla o equilíbrio entre exploration (ações aleatórias) e exploitation (ações otimizadas).

---

#### **2. Inicialização**:
- **Replay Buffer ($D$)**: Comece com um buffer vazio.  
- **Parâmetros da Q-Network ($\theta$)**: Inicialize os pesos da rede principal.  
- **Rede Alvo ($\hat{Q}$)**: Inicialize como uma cópia da rede principal ($\hat{\theta} \leftarrow \theta$).

---

#### **3. Laço de Treinamento**:
- **Laço Externo (Episódios)**: Itere sobre os episódios.  
  1. Reinicie o ambiente e receba a observação inicial $O_0$.  
  2. Crie a sequência inicial $S_0 = \{O_0\}$ e pré-processe para $\phi_0 = \phi(S_0)$.

- **Laço Interno (Passos de Tempo)**:
  - **Passo 1: Seleção de Ação**:
    - Com probabilidade $\epsilon$, selecione uma ação aleatória ($A_t$) para explorar.  
    - Caso contrário, selecione a ação que maximiza o Q-valor:  
      $$
      A_t = \text{argmax}_a Q(\phi(S_t), a; \theta)
      $$

  - **Passo 2: Executar Ação**:
    - Execute a ação $A_t$ no ambiente.  
    - Observe a próxima observação $O_{t+1}$ e a recompensa $R_t$.

  - **Passo 3: Verificar Terminação**:
    - Se o episódio terminar, defina $D_t = 1$ (indicador de término). Caso contrário, $D_t = 0$.

  - **Passo 4: Atualizar Estado**:
    - Atualize a sequência de estados $S_{t+1} = \{S_t, A_t, O_{t+1}\}$.  
    - Pré-processe para $\phi_{t+1} = \phi(S_{t+1})$.

  - **Passo 5: Armazenar Transição**:
    - Armazene a transição $(\phi_t, A_t, R_t, D_t, \phi_{t+1})$ no replay buffer $D$.

  - **Passo 6: Amostrar Mini-Batch**:
    - Amostre um mini-batch aleatório de transições $(\phi_i, A_i, R_i, D_i, \phi'_i)$ de $D$.

  - **Passo 7: Calcular Alvo**:
    - Para cada transição amostrada:  
      - Se $D_i = 0$ (não terminal):  
        $$
        Y_i = R_i + \gamma \max_{a'} Q(\phi'_i, a'; \hat{\theta})
        $$
      - Caso contrário (terminal):  
        $$
        Y_i = R_i
        $$

  - **Passo 8: Descida de Gradiente**:
    - Realize uma etapa de descida de gradiente para minimizar a perda:  
      $$
      L(\theta) = \mathbb{E} \left[(Y_i - Q(\phi_i, A_i; \theta))^2 \right]
      $$

  - **Passo 9: Atualizar Rede Alvo**:
    - A cada $C$ passos, sincronize os parâmetros da rede alvo ($\hat{\theta} \leftarrow \theta$).

  - **Passo 10: Finalizar Episódio**:
    - Se o episódio terminar, saia do laço interno.

---

#### **4. Fim do Treinamento**:
- Continue iterando pelos episódios até alcançar a convergência ou atingir o número máximo de episódios.

---

### **Pontos-Chave**:
1. **Replay Buffer**: Melhora a eficiência dos dados e reduz correlações entre amostras consecutivas.  
2. **Rede Alvo**: Estabiliza o treinamento ao fornecer alvos fixos por um período de tempo ($C$).  
3. **Política $\epsilon$-Greedy**: Garante a exploração do ambiente enquanto explora a política aprendida.  
4. **Pré-processamento**: Empilha e processa os quadros para capturar informações temporais.  
5. **Função de Perda**: Garante que a Q-network se aproxime dos Q-valores ótimos através da descida de gradiente.
