As duas famílias de métodos — **Policy Gradient (PG)** e **Q-Learning** — são algumas das abordagens mais importantes em **Reinforcement Learning (RL)**. Embora ambas tenham o objetivo de aprender um comportamento que maximize a recompensa acumulada, elas fazem isso de maneiras bastante diferentes.

---

# A ideia central

Imagine que você está treinando um robô para jogar um videogame.

Existem duas perguntas possíveis:

### Abordagem 1: "Qual ação tem mais valor?"

Essa é a lógica do **Q-Learning**.

O agente aprende uma função: Q(s,a)

que representa:

> "Se eu estiver no estado *s* e executar a ação *a*, qual recompensa total espero obter no futuro?"

Depois de aprender essa tabela/função, o agente simplesmente escolhe a ação com maior valor Q.

---

### Abordagem 2: "Qual ação devo escolher diretamente?"

Essa é a lógica dos **Policy Gradient Methods**.

Em vez de aprender valores das ações, o agente aprende uma política:

$$\pi(a|s)$$

que representa:

> "Dado o estado *s*, qual a probabilidade de executar cada ação possível?"

Ou seja, ele aprende diretamente o comportamento desejado.

---

# Analogia simples

Suponha que você está escolhendo um restaurante.

## Q-Learning

Você cria uma planilha:

| Restaurante | Nota |
| ----------- | ---- |
| A           | 7.2  |
| B           | 9.1  |
| C           | 8.5  |

Depois escolhe sempre o restaurante com maior nota.

---

## Policy Gradient

Você aprende diretamente uma preferência:

| Restaurante | Probabilidade |
| ----------- | ------------- |
| A           | 5%            |
| B           | 80%           |
| C           | 15%           |

A política já informa qual ação tomar.

# Como o Q-Learning funciona

O algoritmo aprende estimativas dos valores Q.

A atualização clássica do Q-Learning é:

$$Q(s,a)\leftarrow Q(s,a)+\alpha\left[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\right]$$

Onde:

* **r** = recompensa recebida
* **γ** = fator de desconto
* **α** = taxa de aprendizado
* **s'** = próximo estado

A ideia é:

> "Ajuste sua estimativa atual usando uma estimativa melhor do futuro."

---

## Características do Q-Learning

### Vantagens

✅ Alta eficiência amostral (usa bem os dados)

✅ Relativamente estável em problemas discretos

✅ Mais fácil de treinar em ambientes pequenos

✅ Base de algoritmos famosos como:

* Deep Q-Network (DQN)
* Double DQN
* Rainbow DQN

---

### Limitações

❌ Difícil para ações contínuas

Exemplo:

* acelerar 23,17%
* virar 1,4°
* frear 7,8%

Há infinitas possibilidades.

Nesse caso, calcular:

Q(s,a)

para todas as ações torna-se impraticável.

# Como Policy Gradient funciona

Agora o agente possui parâmetros $θ$ que definem a política.

A política é representada como:

$$\pi_\theta(a|s)$$

O objetivo é maximizar a recompensa esperada.

Os parâmetros são ajustados seguindo o gradiente:

$$\nabla_\theta J(\theta)$$

A interpretação é:

> "Mova os parâmetros na direção que aumenta a recompensa."

---

## Intuição

Suponha que uma sequência de ações gerou um resultado excelente.

O algoritmo aumenta a probabilidade dessas ações.

Se o resultado foi ruim:

* diminui a probabilidade delas.

É semelhante ao aprendizado por tentativa e erro.

---

# Características dos Policy Gradients

### Vantagens

✅ Funcionam naturalmente com ações contínuas

✅ Produzem políticas estocásticas

✅ Escalam melhor para problemas complexos

✅ Muito usados em robótica

✅ Base dos algoritmos modernos:

* REINFORCE
* Trust Region Policy Optimization
* Proximal Policy Optimization
* Deep Deterministic Policy Gradient
* Soft Actor-Critic

---

### Limitações

❌ Alta variância

Os gradientes podem ser muito ruidosos.

❌ Precisam de mais experiências para aprender.

❌ Treinamento pode ser instável.

# Off-policy vs On-policy

Essa é outra diferença importante.

## Q-Learning

É normalmente **off-policy**.

O agente pode aprender observando ações diferentes da política atual.

Exemplo:

* coleta dados usando exploração aleatória
* aprende a política ótima usando esses dados

Isso melhora a eficiência amostral.

---

## Policy Gradient puro

Normalmente é **on-policy**.

Aprende usando dados gerados pela própria política atual.

Consequência:

* precisa coletar novas experiências frequentemente.

# Actor-Critic: o melhor dos dois mundos

Na prática, muitos algoritmos modernos combinam as duas ideias.

Temos:

### Actor

Aprende a política:

$$\pi_\theta(a|s)$$

### Critic

Aprende o valor:

$$Q(s,a);\text{ou};V(s)$$

O critic avalia as ações e orienta o actor.

Algoritmos famosos dessa família:

* Advantage Actor-Critic
* Asynchronous Advantage Actor-Critic
* Soft Actor-Critic
* Deep Deterministic Policy Gradient

# Resumo comparativo

| Aspecto             | Q-Learning                               | Policy Gradient          |
| ------------------- | ---------------------------------------- | ------------------------ |
| Aprende             | Valor da ação Q(s,a)                     | Política π(a|s)          |
| Escolha da ação     | argmax Q                                 | Amostragem da política   |
| Espaço discreto     | Excelente                                | Bom                      |
| Espaço contínuo     | Difícil                                  | Excelente                |
| Eficiência amostral | Alta                                     | Menor                    |
| Variância           | Baixa                                    | Alta                     |
| Estabilidade        | Geralmente maior                         | Pode ser instável        |
| Exploração          | Adicionada externamente (ε-greedy, etc.) | Naturalmente estocástica |
| Exemplos            | DQN, Double DQN                          | PPO, TRPO, REINFORCE     |

A forma mais intuitiva de lembrar é:

* **Q-Learning pergunta:** *"Qual ação tem o maior valor?"*
* **Policy Gradient pergunta:** *"Qual política gera as maiores recompensas?"*

O Q-Learning aprende primeiro a **avaliar ações** para depois agir. Já os métodos de Policy Gradient aprendem **diretamente como agir**, ajustando a política para aumentar a recompensa esperada. Isso torna os métodos de Policy Gradient especialmente poderosos em ambientes com ações contínuas e de alta dimensão, enquanto o Q-Learning costuma ser mais eficiente em problemas com ações discretas.


Uma forma muito didática de entender a diferença é implementar os dois métodos no mesmo problema simples.

Vamos usar um ambiente extremamente pequeno:

* Estado único
* Duas ações:

  * Ação 0 → recompensa = 1
  * Ação 1 → recompensa = 0

A política ótima é escolher sempre a ação 0.

---

## 1. Q-Learning

O Q-Learning aprende o valor de cada ação.

````python
import random

# Q-values iniciais
Q = [0.0, 0.0]

alpha = 0.1      # taxa de aprendizado
epsilon = 0.2    # exploração

for episode in range(100):

    # Estratégia epsilon-greedy
    if random.random() < epsilon:
        action = random.randint(0, 1)
    else:
        action = Q.index(max(Q))

    # Ambiente
    reward = 1 if action == 0 else 0

    # Atualização Q-Learning
    Q[action] = Q[action] + alpha * (reward - Q[action])

    if episode % 10 == 0:
        print(f"Ep {episode:3d} | Q = {Q}")

print("\nPolítica aprendida:")
print("Escolher ação:", Q.index(max(Q)))
````

### O que está acontecendo?

O agente aprende:

```text
Q(ação 0) ≈ 1
Q(ação 1) ≈ 0
```

Depois escolhe:

```python
argmax(Q)
```

ou seja:

```python
action = Q.index(max(Q))
```

A política é derivada dos valores Q.

# 2. Policy Gradient (REINFORCE simplificado)

Agora não existe tabela Q.

Aprendemos diretamente a probabilidade de escolher cada ação.

````python
import numpy as np

theta = 0.0      # parâmetro da política
alpha = 0.1

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for episode in range(100):

    # política
    p_action0 = sigmoid(theta)

    # amostra ação
    action = np.random.choice([0, 1],
                              p=[p_action0,
                                 1 - p_action0])

    reward = 1 if action == 0 else 0

    # gradiente do log da política
    if action == 0:
        grad = 1 - p_action0
    else:
        grad = -p_action0

    # atualização
    theta += alpha * reward * grad

    if episode % 10 == 0:
        print(
            f"Ep {episode:3d}"
            f" | P(action0)={p_action0:.3f}"
            f" | P(action1)={1-p_action0:.3f}"
        )

print("\nProbabilidade final:")
print(sigmoid(theta))
````

---

# O que o Policy Gradient está aprendendo?

Não aprende:

```text
Q(ação 0)
Q(ação 1)
```

Aprende diretamente:

```text
P(ação 0)
P(ação 1)
```

Por exemplo:

```text
P(ação 0) = 0.98
P(ação 1) = 0.02
```

# A diferença essencial em uma linha

### Q-Learning

```python
rede -> valores Q -> escolher maior
```

### Policy Gradient

```python
rede -> probabilidades -> amostrar ação
```

Essa é a distinção fundamental que dá origem a quase todos os algoritmos modernos: **DQN aprende valores e deriva a política; REINFORCE/PPO aprendem a política diretamente.**
